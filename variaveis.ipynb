```python
import pandas as pd
import numpy as np
```

```python
dados = {'Alunos': ['Ana', 'Bruno', 'Carol', 'Diego', 'Elisa', 'Felipe', 'Gabi', 'Hugo'],
        'Notas': [8.5, 6.0, 9.2, 5.5, 7.8, 4.0, 9.0, 6.5],
        'Horas de estudo': [6,3,8,2,5,1,7,4]
}
```

```python
df = pd.DataFrame(dados)
df
```

```python
array_nota = np.array(df['Notas'])
array_hora = np.array(df['Horas de estudo'])
```

```python
variancia = np.var(array_nota)
print(f'Variância: {variancia:.2f}')
```

```python
desvio = np.std(df['Notas'])
print(f'Desvio Padrão: {desvio:.2f}')
media = np.mean(df['Notas'])
print(f'Média das notas: {media:.2f}')
print(f'Faixa típica: {media-desvio:.2f} até {media+desvio:.2f}')
```

```python
matriz_cov = np.cov(array_nota, array_hora)
print(matriz_cov)

cov_valor = matriz_cov[0,1]
print(f'Covariancia: {cov_valor:.2f}')
```

```python
print(df[['Notas','Horas de estudo']].cov())
```

```python
r = np.corrcoef(array_nota, array_hora)
print(f'Pearson r = {r[0,1]:.4f}')
```

```python
corr_pandas = df[['Notas', 'Horas de estudo']].corr()
print(corr_pandas)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(13,8))
plt.scatter (array_hora, array_nota, color='steelblue', edgecolor='black', s=80)
plt.title('Horas de Estudo vs Nota')
plt.xlabel('Horas de Estudo')
plt.ylabel('Nota')
plt.show()
```
