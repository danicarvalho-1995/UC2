### Estatísticas de furto de celular por CISP
Agrupa os registros por CISP e soma os casos de furto de celular. Em seguida, calcula medidas estatísticas, a distância relativa entre média e mediana e os limites para identificação de possíveis outliers pelo método do IQR.


```python
df_furto_cel_dp = df.groupby('cisp')['furto_celular'].sum().reset_index()
df_furto_cel_dp = df_furto_cel_dp.sort_values(by='furto_celular', ascending = False)

array_furto_cel_dp = np.array(df_furto_cel_dp['furto_celular'])
array_furto_cel_dp

minimo = np.min(array_furto_cel_dp)
maximo = np.max(array_furto_cel_dp)
amplitude = maximo - minimo
media = np.mean(array_furto_cel_dp)
mediana = np.median(array_furto_cel_dp)
q1 = np.percentile(array_furto_cel_dp, 25)
q2 = np.percentile(array_furto_cel_dp, 50)
q3 = np.percentile(array_furto_cel_dp, 75)
distancia = (media - mediana) / mediana
iqr = q3 - q1
limite_superior = q3 + (1.5 * iqr)
limite_inferior = q1 - (1.5 * iqr)

print(f'Média {media}')
print(f'Médiana {mediana}')
print(f'Q1 {q1}')
print(f'Q2 {q2}')
print(f'Q3 {q3}')
print(f'Distância {distancia:.2%}')
print(f'Limite superior {limite_superior}')
print(f'Limite inferior {limite_inferior}')
```

### Boxplot de furto de celular
Cria um boxplot horizontal para visualizar a distribuição dos furtos de celular entre as CISP, destacando também a média.


```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15, 10))
plt.boxplot(array_furto_cel_dp, orientation='horizontal', showmeans=True)
plt.show
```

### Identificação de outliers
Filtra as CISP cujo total de furtos de celular ultrapassa o limite superior calculado pelo método do IQR.


```python
outlier_furto_cel_dp = df_furto_cel_dp[df_furto_cel_dp['furto_celular'] > limite_superior]

outlier_furto_cel_dp
```

### Identificação de outliers em furto de veículo
Calcula medidas estatísticas e os limites do IQR para os furtos de veículos. Em seguida, identifica os outliers superiores e inferiores.


```python
df_furto_veiculo_bf_dp = df_bf.groupby('cisp')['furto_veiculos'].sum().reset_index().sort_values(by='furto_veiculos', ascending = False)
array_furto_bf = np.array(df_furto_veiculo_bf_dp['furto_veiculos'])


minimo = np.min(array_furto_bf)
maximo = np.max(array_furto_bf)
amplitude = maximo - minimo
media_furto = np.mean(array_furto_bf)
mediana_furto = np.median(array_furto_bf)
q1_furto = np.percentile(array_furto_bf, 25)
q2_furto = np.percentile(array_furto_bf, 50)
q3_furto = np.percentile(array_furto_bf, 75)
iqr = q3_furto - q1_furto
limite_superior = q3_furto + (1.5 * iqr)
limite_inferior = q1_furto - (1.5 * iqr)

print(f'Média: {media_furto}')
print(f'Mediana: {mediana_furto}')
print(f'Q1: {q1_furto}')
print(f'Q3: {q3_furto}')
print(f'Limite superior: {limite_superior}')
print(f'Limite inferior: {limite_inferior}')

outlier_sup = df_furto_veiculo_bf_dp.query("furto_veiculos > @limite_superior")
outlier_inf = df_furto_veiculo_bf_dp.query("furto_veiculos < @limite_inferior")
outlier_sup = outlier_sup.sort_values(by='furto_veiculos', ascending=False)
outlier_inf = outlier_inf.sort_values(by='furto_veiculos', ascending=True)
```
