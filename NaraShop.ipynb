```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('pedidos_narashop.csv', encoding= 'latin1', sep=';')
df
```

```python
df['valor_pedido'] = df['quantidade'] * df['valor_unitario']
df
```

```python
df_valor_pedido = df.groupby('categoria')['valor_pedido'].sum().reset_index()
df_valor_pedido = df_valor_pedido.sort_values(by='valor_pedido', ascending=False)
df_valor_pedido.head()

```

```python
array_total = np.array(df['valor_pedido'])
media_total = np.mean(array_total)
mediana_total = np.median(array_total)
dist = (media_total - mediana_total) / mediana_total
distancia_total = round(((media_total - mediana_total) / mediana_total) * 100,2)

print(f'Média total {media_total}')
print(f'Mediana total {mediana_total}')
print(f'A distância entre a média e a mediana é de {distancia_total}%')

print(dist)
print(f'{dist:.2%}')
```

```python
array_valor_vendido = np.array(df_produto_valor_vendido['valor_pedido'])

q1_valor = np.percentile(array_valor_vendido, 25)
q3_valor = np.percentile(array_valor_vendido, 75)
iqr = q3_valor - q1_valor
limite_superior = q3_valor + (1.5 * iqr)
limite_inferior = q1_valor - (1.5 * iqr)

outlier_sup = df_produto_valor_vendido.query("valor_pedido > @limite_superior")
outlier_inf = df_produto_valor_vendido.query("valor_pedido < @limite_inferior")
outlier_sup = outlier_sup.sort_values(by='valor_pedido', ascending=False)
outlier_inf = outlier_inf.sort_values(by='valor_pedido', ascending=True)

display(outlier_sup)
display(outlier_sup)
```

```python
df_vendedor = df.groupby('vendedor')['valor_pedido'].sum().reset_index().sort_values(by='valor_pedido', ascending=False)
array_vendedor = np.array(df['vendedor'])

q1_vendedor = np.percentile(array_vendedor, 25)
q3_vendedor = np.percentile(array_vendedor, 75)
iqr = q3_vendedor - q1_vendedor
limite_superior = q3_vendedor + (1.5 * iqr)
limite_inferior = q1_vendedor - (1.5 * iqr)

outlier_sup = df_vendedor.query("valor_pedido > @limite_superior")
outlier_inf = df_vendedor.query("valor_pedido < @limite_inferior")
outlier_sup = outlier_sup.sort_values(by='valor_pedido', ascending=False)
outlier_inf = outlier_inf.sort_values(by='valor_pedido', ascending=True)

display(outlier_sup)
display(outlier_sup)
```
